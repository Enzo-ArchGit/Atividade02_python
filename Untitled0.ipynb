```python
livros = []
usuarios = []
emprestimos = []


def cad_livro():
    print("CADASTRAR LIVROS")

    titulo = str(input("Digite o titulo do livro: "))
    autor = str(input("Digite o autor do livro: "))

    livro = {
        "titulo": titulo,
        "autor": autor,
        "disponivel": True
    }

    livros.append(livro)

    print("Livro cadastrado!")


def list_livros():
    print(" LIVROS ")

    if len(livros) == 0:
        print("Nenhum livro cadastrado.")
        return

    for livro in livros:

        if livro["disponivel"]:
            situacao = "Disponível"
        else:
            situacao = "Emprestado"

        print(
            livro["titulo"],"-",
            livro["autor"],"-",
            situacao
        )


def cad_users():
    print("\n CADASTRAR USUÁRIO ")

    nome = input("Digite o nome do usuário: ")

    usuarios.append(nome)

    print("Usuário cadastrado!")


def list_users():
    print("\n USUÁRIOS ")

    if len(usuarios) == 0:
        print("Nenhum usuário cadastrado.")
        return

    for usuario in usuarios:
        print(usuario)


def emprest_livro():

    print("\n EMPRESTAR LIVRO ")

    nome = input("Digite o nome do usuário: ")
    titulo = input("Digite o título do livro: ")


    if nome not in usuarios:
        print("Usuário não encontrado.")
        return


    for livro in livros:

        if livro["titulo"] == titulo:


            if livro["disponivel"] == False:
                print("Esse livro já está emprestado.")
                return


            livro["disponivel"] = False

            emprestimos.append({
                "usuario": nome,
                "livro": titulo
            })

            print("Empréstimo realizado!")
            return

    print("Livro não encontrado.")


def dev_livro():

    print("\n DEVOLVER LIVRO ")

    titulo = input("Digite o título do livro: ")

    for livro in livros:

        if livro["titulo"] == titulo:

            livro["disponivel"] = True

            print("Livro devolvido!")
            return

    print("Livro não encontrado.")


def list_emprest():

    print(" EMPRÉSTIMOS ")

    if len(emprestimos) == 0:
        print("Nenhum empréstimo realizado.")
        return

    for emprestimo in emprestimos:

        print(
            "Usuário:",emprestimo["usuario"],
            "| Livro:",emprestimo["livro"]
        )


while True:

    print("\n==========================")
    print("     BIBLIOTECA")
    print("==========================")

    print("1 - Cadastrar livro")
    print("2 - Listar livros")
    print("3 - Cadastrar usuário")
    print("4 - Listar usuários")
    print("5 - Emprestar livro")
    print("6 - Devolver livro")
    print("7 - Listar empréstimos")
    print("0 - Sair")

    opcao = input("Escolha uma opção: ")

    if opcao == "1":
        cad_livro()

    elif opcao == "2":
        list_livros()

    elif opcao == "3":
        cad_users()

    elif opcao == "4":
        list_users()

    elif opcao == "5":
        emprest_livro()

    elif opcao == "6":
        dev_livro()

    elif opcao == "7":
        list_emprest()

    elif opcao == "0":
        print("Programa encerrado.")
        break

    else:
        print("Opção inválida.")
```
