```python
horas = float(input("Digite aqui as horas: "))

if horas <= 2:
  val = 8
elif horas <= 5:
  val = 8 + 3 * (horas-2)
else:
  val = 17 + 2 * (horas-5)

print(f"Com {horas} horas de uso do estacionamento, o valor cobrado é {val}!")

```

```python
n = int(input("Digite um número inteiro: "))
m = n
p = 1

for i in range(2,9):
  n = int(input("Digite outro número inteiro: "))
  if n >= m:
    m = n
    p = i

print(f"O maior número foi {m}, encontrado na posição {p}")
```

#8

```python
numero = int(input("digite um numero"))
anterior = numero
consecutivos = False

for i in range(0,7):
  numero = int(input("digite outro numero"))
  if numero == anterior:
    consecutivos = True

  anterior = numero

if consecutivos == False:
  print("Não houveram números consecutivos iguais")
else:
  print("Houveram números consecutivos iguais")
```

```python
saldo =  500
for i in range(0,6):
  print ("====Funções====")
  print("1 -> depósito")
  print("2 -> saque ")
  val = float(input("Qual o valor da operação : "))
  op = int(input("Operação escolhida: "))
  if op == 1:
    saldo += val
  elif op == 2 :
    saldo = saldo - op
  else :
    print ("operação invalida ")
print (saldo)

```
